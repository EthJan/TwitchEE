# Two-tower candidate generation

Run `00_get_data.ipynb` first with a Twitch user access token that can read follows. This experiment trains separate user and stream encoders and ranks the live snapshot by embedding dot product. It writes `data/two_tower_top_50.jsonl`.

The only positive labels available here are live channels the user follows. Sampled non-followed channels are unlabeled negatives, not confirmed dislikes. Twitch's `/videos` data is the user's published videos, not their watch history, so it is deliberately not treated as positive feedback. This is a weak-label proof of concept for one user, not a production or behavior-trained recommender. There is no held-out user-feedback evaluation yet.

The user tower embeds followed channels/categories and recently watched channels/categories. The item tower embeds channel, category, language, and tags, concatenated with normalized viewer-count, stream-age, tag-count, and title-length features. Both towers produce normalized 32-dimensional vectors and train with sampled binary cross-entropy. Watch sessions from the signed-in app are aggregated over the last 30 days; longer viewing raises the match for that streamer and category, with watch-time affinity contributing 55% of the final score when history is available.

For this snapshot the exact top-K is computed with a matrix-vector dot product. At larger catalog sizes, item vectors can be precomputed and indexed with an approximate-nearest-neighbor service.
